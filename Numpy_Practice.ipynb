# 1. NumPy

NumPy stands for Numerical Python.

It is a Python library used to perform mathematical and numerical operations efficiently.

#### Ex: Think of like you have 5 employees salaries

```python
salaries = [20000, 25000, 30000, 35000, 40000]
```

#### Requirement : Increase every employees salary by 10%.

##### With normal python

```python
new_salaries = []

for salary in salaries:
    new_salaries.append(salary * 1.10)

print(new_salaries)
```

##### With numpy

```python
import numpy as np

salaries = np.array([20000, 25000, 30000, 35000, 40000])

new_salaries = salaries * 1.10

print(new_salaries)
```

#### Note: NumPy allows to perform operations on entire arrays without explicitly writing loops.

# 9. Broadcasting

Broadcasting allows NumPy to perform operations between arrays of different shapes when their dimensions are compatible.


It is not just "addition". its a set of rules that allows NumPy to perform arithmetic operations on arrays of different shapes. Broadcasting applies equally to subtraction, multiplication, division, and other element wise operations

```python
import numpy as np

array = np.array([10, 20, 30])

result = array + 5 # NumPy automatically broadcasts the scalar 5 across all elements.

print(result)
```

```python
# Real time example:

# Marks for three students in three subjects.
```

```python
marks = np.array([
    [70, 80, 90],
    [60, 75, 85],
    [80, 90, 95]
])

bonus = np.array([10, 5, 9]) # Each subject gets its corresponding bonus.

final_marks = marks + bonus

print(final_marks)
```

# 10. Boolean indexing and filtering

Boolean indexing is used to filter data based on conditions.

Ex:

```python
# Single Condition

arr = np.array([10, 20, 30, 40, 50, 60])

result = arr[arr >= 60]

print(result)
```

```python
# Multiple Condition

result = arr[(arr > 10) & (arr < 50)]

print(result)
```

# 11. Sorting and searching

```python
arr = np.array([50, 10, 40, 20, 30,70,10])

print(np.sort(arr)) # sort() returns sorted values
```

```python
#argsort() returns the original positions of sorted values.

np.argsort(arr)          # Indices of sorted elements -- Returns the indexes that would sort the array in ascending order
```

```python
# np.argmax() → Maximum value's index.
# np.max() → Maximum value.
    
np.argmax(arr)           # Returns the index of the maximum value

# Array:  [50, 10, 40, 20, 30]
# Index:   0   1   2   3   4
```

```python
# np.argmin() → Minimum value's index
# np.min() → Minimum value

np.argmin(arr)           # Returns the index of the minimum value
```

```python
# Values:  [50, 10, 40, 20, 30]
# Indexes: [0,  1,  2,  3,  4]

# Find the indexes of elements satisfying a condition.

np.where(arr > 25)       # Indices satisfying condition
```

```python
np.unique(arr)           # Unique elements
```

```python
## NumPy needs a datatype that can represent both the numbers and NaN.
## Standard NumPy integer types cannot represent NaN, so NumPy promotes the array to float64
```

```python
# Note:

# 8 bits = 1 byte

# Data type	Memory used

# int32	    4 bytes
# int64	    8 bytes
# float32	4 bytes
# float64	8 bytes
```

```python
import numpy as np

a = np.array([10, 20, 30], dtype=np.int32)
b = np.array([10, 20, 30], dtype=np.int64)

print(a)
print(a.dtype)
print(a.nbytes)

print(b)
print(b.dtype)
print(b.nbytes)
```

```python
# int32
# 4 bytes
# Smaller integer range

# int64
# 8 bytes
# Larger integer range
```

```python
# float32
# 4 bytes
# Less decimal precision

# float64
# 8 bytes
# More decimal precision  -- float64 generally provides around 15–16 significant decimal digits
```
